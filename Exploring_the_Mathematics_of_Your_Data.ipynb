```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset (using the Rwanda Financial Inclusion dataset from Week 3)
url = "https://raw.githubusercontent.com/datasets/financial-inclusion/master/RwandaFn.csv"
try:
    df = pd.read_csv(url)
except:
    # Fallback to sample data if URL fails
    df = pd.DataFrame({
        'household_size': np.random.randint(1, 10, 1000),
        'age_of_respondent': np.random.randint(16, 85, 1000),
        'bank_account': np.random.choice(['Yes', 'No'], 1000)
    })

# Display the first few rows to verify
display(df.head())
```

```python
# 1. Select 2 numerical variables
num_cols = ['household_size', 'age_of_respondent']

# 2. Calculate mean, median, range, variance, and standard deviation
stats_dict = {
    'Mean': df[num_cols].mean(),
    'Median': df[num_cols].median(),
    'Range': df[num_cols].max() - df[num_cols].min(),
    'Variance': df[num_cols].var(),
    'Standard Deviation': df[num_cols].std()
}

stats_df = pd.DataFrame(stats_dict)
print("--- Descriptive Statistics ---")
display(stats_df)
```

```python
# 3. Calculate Covariance and Correlation
covar = df['household_size'].cov(df['age_of_respondent'])
corr = df['household_size'].corr(df['age_of_respondent'])

print(f"Covariance between Household Size and Age: {covar:.2f}")
print(f"Correlation between Household Size and Age: {corr:.2f}\n")

# 4. Create a simple plot
plt.figure(figsize=(8, 5))
sns.scatterplot(x='age_of_respondent', y='household_size', data=df, alpha=0.5, color='blue')
plt.title('Relationship between Respondent Age and Household Size')
plt.xlabel('Age of Respondent (Years)')
plt.ylabel('Household Size')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### Statistical Analysis & Explanations

**What do the statistics tell you about your data?**
The descriptive statistics provide a summary of the central tendency and dispersion of our demographic features. The mean and median reveal the average and middle-point age and household size of the respondents, while the variance, standard deviation, and range highlight how spread out these numbers are across the Rwandan population surveyed. A large standard deviation in age indicates a wide demographic reach, from youth to the elderly.

**What does the correlation tell you about the relationship between the two variables?**
The correlation coefficient quantifies the strength and direction of the linear relationship between `age_of_respondent` and `household_size`. A value close to 0 indicates a very weak or non-existent linear relationship, meaning older respondents do not reliably have larger or smaller households than younger respondents.

**Does correlation necessarily mean that one variable causes the other?**
No, correlation does not imply causation. Even if a strong mathematical correlation existed between age and household size, it does not prove that aging *causes* a household to grow or shrink. Underlying confounding variables (like cultural practices, economic shifts, or urban migration) often influence both variables simultaneously without a direct cause-and-effect link between the two.

### Mathematical Expression: Input → Model → Output

If we use **Age** to predict whether someone holds a **Bank Account**, the system is mathematically expressed as:

$y = f(x)$

Where:
*   **$x$ (Input):** $\text{Age\_of\_Respondent}$
*   **$f$ (Function/Model):** The machine learning classification algorithm (e.g., Logistic Regression mapping inputs to a probability curve).
*   **$y$ (Output):** The predicted status ($\text{Bank\_Account} \in \{0, 1\}$).

Fully expanded:
$\text{Bank\_Account\_Probability} = f(\text{Age\_of\_Respondent})$
